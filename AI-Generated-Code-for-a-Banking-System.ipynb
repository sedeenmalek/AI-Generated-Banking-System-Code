```python
class Account:
    """Represents a bank account and its core operations."""
    def __init__(self, account_number, holder_name, initial_balance=0.0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = float(initial_balance)
        self.transactions = []  # To maintain transaction history
        self.add_transaction(f"Account opened with initial balance: {initial_balance}")

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.add_transaction(f"Deposit: +{amount}")
            print(f"Successfully deposited {amount}. Current balance: {self.balance}")
            return True
        else:
            print("Error: Deposit amount must be greater than zero.")
            return False

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                self.add_transaction(f"Withdrawal: -{amount}")
                print(f"Successfully withdrew {amount}. Current balance: {self.balance}")
                return True
            else:
                print("Error: Insufficient funds for this withdrawal.")
                return False
        else:
            print("Error: Withdrawal amount must be greater than zero.")
            return False

    def add_transaction(self, description):
        """Adds a transaction record."""
        self.transactions.append(description)

    def show_statement(self):
        """Displays account details and transaction history."""
        print("\n" + "=" * 40)
        print(f" ACCOUNT STATEMENT - Acc No: {self.account_number}")
        print(f" Holder Name: {self.holder_name}")
        print(f" Current Balance: {self.balance}")
        print("-" * 40)
        print("Recent Transactions:")
        for t in self.transactions:
            print(f" - {t}")
        print("=" * 40 + "\n")

class BankSystem:
    """Manages the overall banking system (accounts creation, transfers, searches)."""
    def __init__(self, bank_name):
        self.bank_name = bank_name
        self.accounts = {}  # Dictionary to store accounts (Key: account_number)

    def create_account(self, account_number, holder_name, initial_balance=0.0):
        if account_number in self.accounts:
            print("Error: Account number already exists.")
            return False

        new_acc = Account(account_number, holder_name, initial_balance)
        self.accounts[account_number] = new_acc
        print(f"Account successfully created for {holder_name} with Account No: {account_number}")
        return True

    def get_account(self, account_number):
        return self.accounts.get(account_number, None)

    def transfer(self, from_acc_num, to_acc_num, amount):
        sender = self.get_account(from_acc_num)
        receiver = self.get_account(to_acc_num)

        if not sender:
            print("Error: Sender account not found.")
            return
        if not receiver:
            print("Error: Receiver account not found.")
            return

        print(f"Transferring {amount} from account ({from_acc_num}) to account ({to_acc_num})...")
        if sender.withdraw(amount):
            receiver.deposit(amount)
            sender.add_transaction(f"Outbound transfer of {amount} to {to_acc_num}")
            receiver.add_transaction(f"Inbound transfer of {amount} from {from_acc_num}")
            print("Transfer completed successfully!")

# ==========================================
# Interactive Command Line Interface (CLI)
# ==========================================
def main():
    my_bank = BankSystem("Python International Bank")

    while True:
        print(f"\n--- {my_bank.bank_name} Management System ---")
        print("1. Create New Account")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Transfer Money Between Accounts")
        print("5. View Account Statement")
        print("6. Exit")

        choice = input("Choose an option (1-6): ").strip()

        if choice == '1':
            acc_num = input("Enter Account Number: ").strip()
            name = input("Enter Account Holder Name: ").strip()
            try:
                balance = float(input("Enter Initial Balance (default 0): ") or 0)
                my_bank.create_account(acc_num, name, balance)
            except ValueError:
                print("Invalid balance input.")

        elif choice == '2':
            acc_num = input("Enter Account Number: ").strip()
            acc = my_bank.get_account(acc_num)
            if acc:
                try:
                    amount = float(input("Enter deposit amount: "))
                    acc.deposit(amount)
                except ValueError:
                    print("Please enter a valid numeric amount.")
            else:
                print("Account not found.")

        elif choice == '3':
            acc_num = input("Enter Account Number: ").strip()
            acc = my_bank.get_account(acc_num)
            if acc:
                try:
                    amount = float(input("Enter withdrawal amount: "))
                    acc.withdraw(amount)
                except ValueError:
                    print("Please enter a valid numeric amount.")
            else:
                print("Account not found.")

        elif choice == '4':
            from_acc = input("Enter sender's account number: ").strip()
            to_acc = input("Enter receiver's account number: ").strip()
            try:
                amount = float(input("Enter transfer amount: "))
                my_bank.transfer(from_acc, to_acc, amount)
            except ValueError:
                print("Please enter a valid numeric amount.")

        elif choice == '5':
            acc_num = input("Enter account number to view statement: ").strip()
            acc = my_bank.get_account(acc_num)
            if acc:
                acc.show_statement()
            else:
                print("Account not found.")

        elif choice == '6':
            print("Thank you for using the Banking System. Goodbye!")
            break
        else:
            print("Invalid choice. Please select an option between 1 and 6.")

if __name__ == "__main__":
    main()
```
